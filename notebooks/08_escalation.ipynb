```python
%load_ext autoreload
%autoreload 2

import sqlite3
from bankingsupportagent.graph import build_graph
from bankingsupportagent.config import DB_PATH

graph = build_graph()

def chat(message, thread_id, customer_id=None):
    inputs = {"messages": message}
    if customer_id:
        inputs["customer_id"] = customer_id
    result = graph.invoke(inputs, config={"configurable": {"thread_id": thread_id}})
    print(f"Q: {message}\nRoute: {result['route']}\nA: {result['messages'][-1].content}\n")

chat("Someone logged into my account and transferred Rs 20,000 out without my permission", "e1", "C1001")
chat("I've been waiting 3 weeks for my refund and nobody is helping me, this is terrible service", "e2", "C1002")
chat("I want to talk to a manager", "e3", "C1001")
chat("My card was stolen and someone is using it", "e4")       # not logged in

conn = sqlite3.connect(DB_PATH)
for row in conn.execute("SELECT ticket_id, customer_id, category, priority, summary FROM tickets"):
    print(row)
conn.close()
```
