```python
from bankingsupportagent.rag.retriever import get_vector_store

questions = [
    "Can I freeze my card instead of blocking it?",
    "What is the FD interest rate for a senior citizen for 1 year?",
    "Is the branch open on Saturday?",
    "How many days do I have to report a fraud transaction?",
    "What is the interest rate on home loans?",
]

store = get_vector_store()
for q in questions:
    print(f"\nQ: {q}")
    for doc, score in store.similarity_search_with_score(q, k=3):
        print(f"   {score:.3f} | {doc.metadata['source']} p{doc.metadata['page']} | {doc.page_content[:70]!r}")
```
