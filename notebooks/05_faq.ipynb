```python
%load_ext autoreload
%autoreload 2

from bankingsupportagent.graph import build_graph

graph = build_graph()

def ask(question, thread_id="test"):
    result = graph.invoke(
        {"messages": question},
        config={"configurable": {"thread_id": thread_id}},
    )
    print(f"Q: {question}")
    print(f"Route: {result['route']}")
    print(f"A: {result['messages'][-1].content}\n")
```

```python
ask("Can I freeze my card instead of blocking it?", "t1")
ask("What is the FD interest rate for a senior citizen for 1 year?", "t2")
ask("Is the branch open on Saturday?", "t3")
ask("What is the interest rate on home loans?", "t4")
ask("Should I invest in an FD or in mutual funds?", "t5")
```
